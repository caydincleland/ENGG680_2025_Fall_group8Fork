## Part 6: Linear LSE 
### (35 marks)

### Curve Fitting

curvefitting.png

### Observations: values of a function f(a) in range of -π to π, contaminated with random noise

### Unknown States: coefficients of polynomial functions for curve fitting

```python
# step 1
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# step 2
# Load data.xlsx file, which contains the observations (with noise), variance of observations and true observations
# hint: use pandas library to parse excel file and use numpy library to convert it to array
# (1 mark)
df_data = pd.read_excel('data.xlsx')
arr_data = df_data.to_numpy()

#store the noisy observations, variance of observations and true observations into numpy arrays seperately
# (1 mark)
l = arr_data[:, 0]
l_var = arr_data[:, 1]
l_true = arr_data[:, 2]
```

```python
# step 3
# define the number of observation n
# (1 mark)
n = len(l)

# define the observation weight matrix P for LSE as identity matrix
# identity weight matrix simply means the observations will be unweighted in the following LSE
# (1 mark)
P = np.identity(n)

# define the range of function input(i.e., constants a), from start to end; and the numerical array of a
start = -1* np.pi
end = np.pi
a = np.linspace(start, end, n)
```

```python
# step 4
# use the matplotlib to plot the true and noisy observations together in one single figure
# use scatter for noisy observation plotting
# for the true observation, label it as "f(a)"" and set color as 'k'
# for the noisy observation scatter plot, label it as "f(a) + e", set color as 'tab:blue', set marker as 'o' and set alpha as 0.3
# include proper axis labels, legend and title
# scale the range of x-axis to be -1 to 1
# keep this format for the rest of the plots in this lab assignment
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

# Plot true observations
ax.plot(a, l_true, label="f(a)", color='k')

# Plot noisy observations
ax.scatter(a, l, label="f(a) + e", color='tab:blue', marker='o', alpha=0.3)

# Labels, legend, and title
ax.set_xlabel("a (radians)")
ax.set_ylabel("Observation value")
ax.set_title("True vs Noisy Observations")
ax.legend()

# Limit x-axis range to -1 to 1
ax.set_xlim(-1, 1)

ax.grid(True, which='both',ls='-',color='0.65')
fig.tight_layout()
plt.show()
```

```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterier variance factor (goodness of fit)
        v:              observation residuals
    '''

  # === 1. Construct the design matrix A ===
    # Each column corresponds to a^power (starting from degree 0 up to deg)
    A = np.vander(a, N=deg + 1, increasing=True)

    # === 2. Perform Linear Least Squares Estimation (LSE) ===
    # Normal equation: x = (Aᵀ P A)⁻¹ Aᵀ P l
    N = A.T @ P @ A                 # Normal matrix
    U = A.T @ P @ l                 # Right-hand side vector
    x = np.linalg.inv(N) @ U        # Estimated polynomial coefficients

    # === 3. Compute residuals ===
    l_hat = A @ x                   # Fitted (predicted) observations
    v = l - l_hat                   # Residuals = observed - predicted

    # === 4. Compute a posterior variance factor (goodness of fit) ===
    # apv = (vᵀ P v) / (n - m), where m = number of parameters
    n = len(l)
    m = len(x)
    apv = float((v.T @ P @ v) / (n - m))

    # === 5. Compute covariance matrix of estimated coefficients ===
    cov_x = apv * np.linalg.inv(N)

    return (x, cov_x, apv, v)
```

```python
# step 6
# call the fit_polynomial_curve function and fetch the results
# use polynomial degrees of: 2, 4, 6
# (1 marks)
results = {}  # store results for each degree

for deg in [2, 4, 6]:
    x, cov_x, apv, v = fit_polynomial_curve(a, l, P, deg)
    results[deg] = {
        'x': x,
        'cov_x': cov_x,
        'apv': apv,
        'v': v
    }

# Optionally print summaries
for deg, res in results.items():
    print(f"Degree {deg}:")
    print(f"  Coefficients: {res['x']}")
    print(f"  A posterior variance factor: {res['apv']:.4f}")
    print()
```

```python
# step 7
# use the estimated coefficients x and constants array a, construct polynomial functions for each degree
# hint: use polyval function from numpy library (check input argument carefully)
# (1 marks)

quadratic_uw = np.polyval(results[2]['x'][::-1], a)   # degree 2
quartic_uw  = np.polyval(results[4]['x'][::-1], a)    # degree 4
hexic_uw    = np.polyval(results[6]['x'][::-1], a)    # degree 6
```

```python
# step 8
# use the matplotlib to plot each polynomial function, noisy observations l, true observations l_true, together in one single figure
# for the polynomials of line, cubic and quintic, use color of orange, green and red respectively.
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))
# Plot noisy observations (scatter)
ax.scatter(a, l, label="f(a) + e (Noisy)", color='tab:blue', marker='o', alpha=0.3)

# Plot true observations (black line)
ax.plot(a, l_true, label="f(a) (True)", color='k')

# Plot fitted polynomial curves
ax.plot(a, quadratic_uw, label="Quadratic (deg=2)", color='orange')
ax.plot(a, quartic_uw,  label="Quartic (deg=4)", color='green')
ax.plot(a, hexic_uw,    label="Hexic (deg=6)", color='red')

# Add labels, legend, and title
ax.set_xlabel("a (radians)")
ax.set_ylabel("Observation value")
ax.set_title("Polynomial Curve Fitting (Unweighted LSE)")
ax.legend()

# Limit x-axis range
ax.set_xlim(-1, 1)
ax.grid(True, which='both',ls='-',color='0.65')
fig.tight_layout()
plt.show()
```

```python
# step 12
# now, let's define the P matrix with the proper weights of observation
# use the variance of observation l_var, define P matrix
# (1 mark)
# Observation weight matrix: P = diag(1 / variance)
P = np.diag(1 / l_var)
```

```python
# step 13
# repeat the steps 6-10, using the proper P matrix
# (7 marks)
# 1) Fit for degrees 2,4,6 using weighted P
weighted_results = {}
for deg in [2, 4, 6]:
    x_w, cov_x_w, apv_w, v_w = fit_polynomial_curve(a, l, P, deg)
    weighted_results[deg] = {
        'x': x_w,
        'cov_x': cov_x_w,
        'apv': apv_w,
        'v': v_w
    }

# 2) Construct polynomial evaluation arrays (np.polyval expects coeffs highest->lowest)
quadratic_w = np.polyval(weighted_results[2]['x'][::-1], a)
quartic_w   = np.polyval(weighted_results[4]['x'][::-1], a)
hexic_w     = np.polyval(weighted_results[6]['x'][::-1], a)

# 3) Plot noisy obs, true function, and weighted fits
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

# Noisy observations
ax.scatter(a, l, label="f(a) + e (Noisy)", color='tab:blue', marker='o', alpha=0.3)

# True observations
ax.plot(a, l_true, label="f(a) (True)", color='k')

# Weighted fitted polynomial curves
ax.plot(a, quadratic_w, label="Quadratic (deg=2) - weighted", color='orange')
ax.plot(a, quartic_w,  label="Quartic (deg=4) - weighted", color='green')
ax.plot(a, hexic_w,    label="Hexic (deg=6) - weighted", color='red')

ax.set_xlabel("a (radians)")
ax.set_ylabel("Observation value")
ax.set_title("Polynomial Curve Fitting (Weighted LSE: P = diag(1/variance))")
ax.legend()
ax.set_xlim(-1, 1)
ax.grid(True, which='both', ls='-', color='0.65')
fig.tight_layout()
plt.show()

# 4) Print a posterior variance factors and residual norms for each degree
print("Weighted LSE: A Posterior Variance Factor and Residual Norms\n")
for deg in [2, 4, 6]:
    apv = weighted_results[deg]['apv']
    v   = weighted_results[deg]['v']
    res_norm = np.linalg.norm(v)                  # ||v||
    res_norm_sq = float(v.T @ v)                  # v^T v
    print(f"Degree {deg}: APV = {apv:.6e}, Residual norm ||v|| = {res_norm:.6e}, v^T v = {res_norm_sq:.6e}")
```
